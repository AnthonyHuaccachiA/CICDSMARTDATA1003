```python
dbutils.widgets.removeAll()
```

```python
dbutils.widgets.text("catalogo","catalog_smartdata")
```

```python
catalogo = dbutils.widgets.get("catalogo")
```

```python
spark.sql(f"DROP CATALOG IF EXISTS {catalogo} CASCADE;")
```

```python
spark.sql(f"CREATE CATALOG IF NOT EXISTS {catalogo};")
```

```python
spark.sql(f"""
CREATE SCHEMA IF NOT EXISTS {catalogo}.raw;
CREATE SCHEMA IF NOT EXISTS {catalogo}.bronze;
CREATE SCHEMA IF NOT EXISTS {catalogo}.silver;
CREATE SCHEMA IF NOT EXISTS {catalogo}.golden;
CREATE SCHEMA IF NOT EXISTS {catalogo}.exploratory;

CREATE VOLUME IF NOT EXISTS {catalogo}.raw.datasets;
""")
```

###Tablas Bronze

```python
spark.sql(f"""
CREATE TABLE {catalogo}.bronze.circuits (
  circuit_id INT,
  circuit_ref STRING,
  name STRING,
  location STRING,
  country STRING,
  latitude DOUBLE,
  longitude DOUBLE,
  altitude INT,
  ingestion_date TIMESTAMP)
USING DELTA;
""")
```

```python
spark.sql(f"""
CREATE TABLE {catalogo}.bronze.races (
  race_id INT,
  race_year INT,
  round INT,
  circuit_id INT,
  name STRING,
  ingestion_date TIMESTAMP)
USING delta
PARTITIONED BY (race_year);
""")
```

###Tablas Silver

```python
spark.sql(f"""
CREATE TABLE {catalogo}.silver.circuits_transformed (
  race_id INT,
  race_year INT,
  round INT,
  circuit_id INT,
  name_race STRING,
  ingestion_date TIMESTAMP,
  circuit_ref STRING,
  name STRING,
  location STRING,
  country STRING,
  latitude double,
  longitude double,
  altitude INT,
  altitude_category STRING,
  years_diferences INT,
  lat_diff INT,
  race_type STRING,
  near_equator STRING)
USING DELTA;
""")
```

###Tablas Golden

```python
spark.sql(f"""
CREATE TABLE {catalogo}.golden.golden_raced_partitioned (
  race_year STRING,
  conteo LONG,
  max_altitude STRING,
  min_altitude STRING,
  country STRING,
  race_type STRING,
  near_equator STRING)
USING DELTA;
""")
```
